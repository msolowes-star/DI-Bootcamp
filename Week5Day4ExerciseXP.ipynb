Exercise 1: Understanding Data Visualization
1. Why is data visualization important in data analysis?

Answer:

Data visualization is important because it helps people understand data more easily. Charts and graphs reveal patterns, trends, relationships, and outliers that may be difficult to notice by looking at raw numbers. Good visualizations also make it easier to communicate findings to others.

2. What is the purpose of a line graph?

Answer:

A line graph is used to show how data changes over time or another continuous variable. It helps identify trends, increases, decreases, and patterns in the data.

Exercise 2: Line Plot

```python
import matplotlib.pyplot as plt

# Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Create plot
plt.figure(figsize=(8,5))
plt.plot(days, temperatures, marker='o')

# Labels and title
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.title("Temperature Variation Over One Week")

# Display
plt.show()
```

Exercise 3: Bar Chart

```python
import matplotlib.pyplot as plt

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [5000, 5500, 6200, 7000, 7500]

# Create bar chart
plt.figure(figsize=(8,5))
plt.bar(months, sales)

# Labels
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.title("Monthly Sales")

# Display
plt.show()
```

Exercise 4: Histogram of CGPA

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="What is your CGPA?",
    color="skyblue"
)

plt.title("Distribution of Student CGPA")
plt.xlabel("CGPA")
plt.ylabel("Count")

plt.show()
```

Exercise 5: Comparing Anxiety Levels Across Different Genders

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Yes/No responses to numbers
df["AnxietyNumeric"] = df["Do you have Anxiety?"].map({
    "Yes": 1,
    "No": 0
})

# Create the bar plot
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Choose your gender",
    y="AnxietyNumeric",
    hue="Choose your gender",
    palette="Set2",
    legend=False,
    errorbar=None
)

plt.title("Proportion of Students with Anxiety by Gender")
plt.xlabel("Gender")
plt.ylabel("Proportion with Anxiety")
plt.ylim(0, 1)

plt.show()
```

```python
from matplotlib.ticker import PercentFormatter

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Choose your gender",
    y="AnxietyNumeric",
    hue="Choose your gender",
    palette="Set2",
    legend=False,
    errorbar=None
)

plt.title("Proportion of Students with Anxiety by Gender")
plt.xlabel("Gender")
plt.ylabel("Students Experiencing Anxiety")
plt.ylim(0, 1)
plt.gca().yaxis.set_major_formatter(PercentFormatter(1))

plt.show()
```

Exercise 6: Exploring the Relationship Between Age and Panic Attacks

```python
df["PanicNumeric"] = df["Do you have Panic attack?"].map({
    "Yes": 1,
    "No": 0
})
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Age",
    y="PanicNumeric",
    alpha=0.6
)

plt.title("Relationship Between Age and Panic Attacks")
plt.xlabel("Age")
plt.ylabel("Panic Attack (0 = No, 1 = Yes)")

plt.show()
```
