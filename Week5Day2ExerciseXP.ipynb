Exercise 2: Descriptive Statistics

Calculate the mean, median, variance, and standard deviation.

```python
import numpy as np
from scipy import stats

data = [12, 15, 13, 12, 18, 20, 22, 21]

mean_value = stats.tmean(data)
median_value = np.median(data)
variance_value = stats.tvar(data)
standard_deviation = stats.tstd(data)

print("Mean:", mean_value)
print("Median:", median_value)
print("Variance:", variance_value)
print("Standard deviation:", standard_deviation)
```

Exercise 3: Understanding Distributions

Generate and plot a normal distribution with a mean of 50 and a standard deviation of 10.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 50
standard_deviation = 10

# Create x-values covering approximately four standard deviations
x = np.linspace(
    mean - 4 * standard_deviation,
    mean + 4 * standard_deviation,
    500
)

# Calculate the probability density for each x-value
y = norm.pdf(x, loc=mean, scale=standard_deviation)

plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.axvline(mean, linestyle="--", label="Mean = 50")
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

Exercise 4: T-Test Application

Generate two samples and test whether their means are significantly different.

```python
import numpy as np
from scipy.stats import ttest_ind

# Makes the randomly generated data reproducible
np.random.seed(42)

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_statistic, p_value = ttest_ind(data1, data2)

print("Mean of data1:", np.mean(data1))
print("Mean of data2:", np.mean(data2))
print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
  print("Reject the null hypothesis.")
  print("The two groups have significantly different means.")
else:
  print("Fail to reject the null hypothesis.")
  print("There is not enough evidence of a significant difference.")

```

Exercise 5: Linear Regression Analysis

Use house size to predict house price.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

regression = linregress(house_sizes, house_prices)

slope = regression.slope
intercept = regression.intercept
r_value = regression.rvalue
p_value = regression.pvalue

print("Slope:", slope)
print("Intercept:", intercept)
print("R-value:", r_value)
print("R-squared:", r_value ** 2)
print("P-value:", p_value)

house_size_to_predict = 90
predicted_price = intercept + slope * house_size_to_predict
print(
    f"Predicted price for a {house_size_to_predict} square meter house:",
    f"{predicted_price:,.2f}"
)

predicted_prices = intercept + slope * house_sizes

plt.figure(figsize=(8, 5))
plt.scatter(house_sizes, house_prices, label="Actual prices")
plt.plot(house_sizes, predicted_prices, label="Regression line")
plt.scatter(
    house_size_to_predict,
    predicted_price,
    marker="x",
    s=100,
    label="90 m² prediction"
)

plt.title("House Size and Price")
plt.xlabel("House Size in Square Meters")
plt.ylabel("House Price")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

```

Exercise 6: Understanding ANOVA

Test whether the mean plant growth differs among the three fertilizers.

```python
from scipy.stats import f_oneway
import numpy as np

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = f_oneway(
    fertilizer_1,
    fertilizer_2,
    fertilizer_3 )

print("Mean growth with fertilizer 1:", np.mean(fertilizer_1))
print("Mean growth with fertilizer 2:", np.mean(fertilizer_2))
print("Mean growth with fertilizer 3:", np.mean(fertilizer_3))
print("F-value:", f_value)
print("P-value:", p_value)

if p_value < 0.05:
  print("Reject the null hypothesis.")
  print("At least one fertilizer has a significantly different effect.")
else:
  print("Fail to reject the null hypothesis.")
  print("There is not enough evidence that the fertilizers differ.")
```

Exercise 7: Probability Distributions

Calculate the probability of getting exactly five heads in ten fair coin flips.

```python
from scipy.stats import binom

number_of_flips = 10
number_of_heads = 5
probability_of_head = 0.5

probability = binom.pmf(
    number_of_heads,
    number_of_flips,
    probability_of_head
)

print("Probability of exactly 5 heads:", probability)
print("Percentage:", probability * 100)

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

number_of_flips = 10
possible_heads = np.arange(0, number_of_flips + 1)

probabilities = binom.pmf(
    possible_heads,
    number_of_flips,
    0.5
)

for heads, probability in zip(possible_heads, probabilities):
  print(f"{heads} heads: {probability:.4f}")

plt.figure(figsize=(8, 5))
plt.bar(possible_heads, probabilities)
plt.title("Binomial Distribution: Heads in 10 Coin Flips")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.xticks(possible_heads)
plt.show()
```

Exercise 8: Correlation Coefficients

Calculate Pearson and Spearman correlations between age and income.

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

data = pd.DataFrame({
    "age": [23, 25, 30, 35, 40],
    "income": [35000, 40000, 50000, 60000, 70000]
})

pearson_coefficient, pearson_p_value = pearsonr(
    data["age"],
    data["income"]
)

spearman_coefficient, spearman_p_value = spearmanr(
    data["age"],
    data["income"]
)

print("Pearson correlation:", pearson_coefficient)
print("Pearson P-value:", pearson_p_value)

print("Spearman correlation:", spearman_coefficient)
print("Spearman P-value:", spearman_p_value)
```
