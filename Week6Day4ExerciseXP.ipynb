Exercise 1 — Matrix Operations

```python
matrix = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])

determinant = np.linalg.det(matrix)
inverse = np.linalg.inv(matrix)

print("Matrix:")
print(matrix)

print("\nDeterminant:")
print(determinant)

print("\nInverse:")
print(inverse)
```

For a grayscale image, each number represents the brightness of one pixel.

```python
image = np.array([
    [0,   50, 100, 150, 200],
    [50, 100, 150, 200, 255],
    [100, 150, 255, 150, 100],
    [150, 200, 150, 100, 50],
    [200, 150, 100, 50, 0]
])

print(image)
```

Exercise 6 — Basic Hypothesis Testing

```python
np.random.seed(42)

productivity_before = np.random.normal(
    loc=50,
    scale=10,
    size=30
)

productivity_after = productivity_before + np.random.normal(
    loc=5,
    scale=3,
    size=30
)
```

Null hypothesis: The training program does not improve employee productivity.

Alternative hypothesis: The training program improves employee productivity.

```python
before_mean = np.mean(productivity_before)
after_mean = np.mean(productivity_after)

print("Average before training:", before_mean)
print("Average after training:", after_mean)
```

```python
differences = productivity_after - productivity_before

average_improvement = np.mean(differences)

print("Average improvement:", average_improvement)
```

```python
employees_improved = np.sum(productivity_after > productivity_before)

print("Employees who improved:", employees_improved)
print("Total employees:", len(productivity_before))
```

The average productivity score after training was higher than the average before training. The average change was positive, suggesting that the training program may have improved employee productivity.

Exercise 8 - Time Series Manupulation
