Hypothesis testing:

Let's compare two common fuel types such as Coal and Gas.

Our hypotheses are:

H₀ (null hypothesis): Coal and gas plants have the same mean capacity.

H₁ (alternative hypothesis): Coal and gas plants have different mean capacities.

```python
from scipy.stats import ttest_ind
```

```python
coal = df.loc[
    df["primary_fuel"] == "Coal",
    "capacity_mw"
].dropna()

gas = df.loc[
    df["primary_fuel"] == "Gas",
    "capacity_mw"
].dropna()
```

```python
print("Coal mean:", coal.mean())
print("Gas mean:", gas.mean())
```

```python
t_stat, p_value = ttest_ind(
    coal,
    gas,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("There is evidence of a significant difference in mean capacity.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is not enough evidence of a significant difference.")
```

Task 4 — Time Series Analysis

```python
year_data = df.dropna(
    subset=["commissioning_year"]
).copy()
```

```python
year_data["commissioning_year"] = (
    year_data["commissioning_year"]
    .astype(int)
)
```

```python
year_data = year_data[
    (year_data["commissioning_year"] >= 1900) &
    (year_data["commissioning_year"] <= 2026)
]
```

```python
plants_by_year = (
    year_data
    .groupby("commissioning_year")
    .size()
)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    plants_by_year.index,
    plants_by_year.values
)

plt.title("Power Plants Commissioned by Year")
plt.xlabel("Year")
plt.ylabel("Number of Power Plants")
plt.grid(True)

plt.show()
```

Fuel mix over time

```python
year_data["decade"] = (
    year_data["commissioning_year"] // 10
) * 10
```

```python
fuel_by_decade = pd.crosstab(
    year_data["decade"],
    year_data["primary_fuel"]
)

fuel_by_decade.tail()
```

```python
major_fuels = [
    "Coal",
    "Gas",
    "Hydro",
    "Solar",
    "Wind"
]

available_fuels = [
    fuel for fuel in major_fuels
    if fuel in fuel_by_decade.columns
]

fuel_by_decade[available_fuels].plot(
    figsize=(12, 7),
    marker="o"
)

plt.title("Power Plant Fuel Types by Decade")
plt.xlabel("Decade")
plt.ylabel("Number of Plants")
plt.grid(True)
plt.legend(title="Fuel Type")

plt.show()
```

Task 6 — Matrix Operations

```python
matrix_df = df[
    [
        "capacity_mw",
        "latitude",
        "longitude"
    ]
].dropna()
```

```python
matrix = matrix_df.to_numpy()

print(matrix.shape)
print(matrix[:5])
```

```python
matrix_standardized = (
    matrix - np.mean(matrix, axis=0)
) / np.std(matrix, axis=0)
```

```python
correlation_matrix = np.corrcoef(
    matrix_standardized,
    rowvar=False
)

print(correlation_matrix)
```

Eigenvalues and eigenvectors

```python
eigenvalues, eigenvectors = np.linalg.eig(
    correlation_matrix
)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```
