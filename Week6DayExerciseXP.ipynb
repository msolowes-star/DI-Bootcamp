```python
import numpy as np
```

Exercise 1 — Array Creation

```python
arr = np.arange(10)
print(arr)
```

Exercise 2 — Type Conversion

```python
numbers = [3.14, 2.17, 0, 1, 2]

arr = np.array(numbers)
arr_int = arr.astype(int)

print(arr_int)
```

Exercise 3 — 3×3 Array

```python
arr = np.arange(1, 10).reshape(3, 3)

print(arr)
```

Exercise 4 — Random 4×5 Array

```python
arr = np.random.random((4, 5))

print(arr)
```

Exercise 5 — Select the Second Row

```python
arr = np.array([
    [21, 22, 23, 22, 22],
    [20, 21, 22, 23, 24],
    [21, 22, 23, 22, 22]
])

second_row = arr[1]

print(second_row)
```

Exercise 6 — Reverse an Array

```python
arr = np.arange(10)

reversed_arr = arr[::-1]

print(reversed_arr)
```

Exercise 7 — Identity Matrix

```python
identity_matrix = np.eye(4)

print(identity_matrix)
```

Exercise 8 — Sum and Average

```python
arr = np.arange(10)

total = np.sum(arr)
average = np.mean(arr)

print("Sum:", total)
print("Average:", average)
```

Exercise 9 — Reshape 1–20

```python
arr = np.arange(1, 21)

matrix = arr.reshape(4, 5)

print(matrix)
```

Exercise 10 — Select Odd Numbers

```python
arr = np.arange(10)

odd_numbers = arr[arr % 2 != 0]

print(odd_numbers)
```
