Import the libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Generate the temperature data

```python
# City names
cities = [
    "New York",
    "London",
    "Paris",
    "Tokyo",
    "Sydney",
    "Toronto",
    "Rome",
    "Berlin",
    "Madrid",
    "Jerusalem"
]

# Month names
months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

# Generate random temperatures between -5 and 35
np.random.seed(42)

temperature_data = np.random.uniform(
    low=-5,
    high=35,
    size=(10, 12)
)

temperature_data
```

Convert the NumPy array into a Pandas DataFrame

```python
weather_df = pd.DataFrame(
    temperature_data,
    index=cities,
    columns=months
)

weather_df
```

Calculate the annual average temperature

```python
annual_average = weather_df.mean(axis=1)

annual_average
```

```python
weather_df["Annual Average"] = annual_average

weather_df
```

Find the hottest and coldest cities

```python
hottest_city = annual_average.idxmax()
coldest_city = annual_average.idxmin()

print("City with highest average temperature:", hottest_city)
print("City with lowest average temperature:", coldest_city)
```

```python
print(
    "Highest average temperature:",
    hottest_city,
    round(annual_average[hottest_city], 2),
    "°C"
)

print(
    "Lowest average temperature:",
    coldest_city,
    round(annual_average[coldest_city], 2),
    "°C"
)
```

Create the main visualization

```python
plt.figure(figsize=(12, 7))

for city in cities:
    plt.plot(
        months,
        weather_df.loc[city, months],
        marker="o",
        label=city
    )

plt.title("Monthly Temperature Trends Across Cities")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)
plt.show()
```

Chart of annual averages

```python
annual_average.sort_values().plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Annual Temperature by City")
plt.xlabel("City")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
